加载本地大模型作为模型和分词器

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
from langchain.llms import HuggingFacePipeline
import torch

# 加载本地的模型和分词器
model_path = "../xcw/langchain/chatglm3-6b"  # 替换为你的本地模型路径
tokenizer = AutoTokenizer.from_pretrained(model_path)
model = AutoModelForCausalLM.from_pretrained(model_path)

# 使用管道加载模型
pipe = pipeline("text-generation", model=model, tokenizer=tokenizer, device=0 if torch.cuda.is_available() else -1)
llm = HuggingFacePipeline(pipeline=pipe)
```

加载本地解析Yang得到的Json文件

```python
import json
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.docstore.document import Document

# 读取 JSON 文件
with open("path/to/your/file.json", "r") as f:
    data = json.load(f)

# 将 JSON 数据转为 Document 格式
documents = [Document(page_content=item["content"], metadata=item) for item in data]

# 可选：将内容分片，便于检索
text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=100)
docs = text_splitter.split_documents(documents)
```

利用向量化模型将文档转化为嵌入向量，并存入向量数据库（例如 FAISS 或 Chroma）中

```python
from langchain.embeddings import HuggingFaceEmbeddings
from langchain.vectorstores import FAISS

# 使用句子嵌入模型
embedding_model = HuggingFaceEmbeddings(model_name="sentence-transformers/all-MiniLM-L6-v2")
vector_store = FAISS.from_documents(docs, embedding_model)
```

构建一个 RAG 管道，首先用向量存储检索最相关的文档，然后用本地模型生成答案

```python
from langchain.chains import RetrievalQA

# 设置检索增强生成（RAG）链
qa_chain = RetrievalQA.from_llm(llm=llm, retriever=vector_store.as_retriever())

# 使用 RAG 进行问答
query = "你的问题"
result = qa_chain.run(query)
print(result)
```
